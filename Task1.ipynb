```python
import pandas as pd

df = pd.read_excel("Friend_Task1_Raw_Sales_Dataset.xlsx")

print("Dataset loaded successfully!")
```

```python
print("===== DATASET SHAPE =====")
print("Rows and Columns:", df.shape)

print("\n===== COLUMN NAMES =====")
print(df.columns.tolist())

print("\n===== FIRST 5 ROWS =====")
display(df.head())

print("\n===== LAST 5 ROWS =====")
display(df.tail())

print("\n===== DATASET INFORMATION =====")
df.info()
```

```python
print("===== MISSING VALUES =====")
print(df.isnull().sum())

print("\n===== DUPLICATE ROWS =====")
print("Number of duplicate rows:", df.duplicated().sum())

print("\n===== UNIQUE VALUES =====")

for column in df.columns:
    print(f"\n{column}:")
    print(df[column].dropna().unique()[:20])
```

```python
print("===== DATA TYPES =====")
print(df.dtypes)

print("\n===== QUANTITY VALUES =====")
print(df["Quantity"].unique())

print("\n===== SALES AMOUNT SAMPLE =====")
print(df["Sales_Amount"].head(20))

print("\n===== ORDER DATE SAMPLE =====")
print(df["Order_Date"].head(10))
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Identify text columns
text_columns = df.select_dtypes(include="object").columns

# Remove extra spaces
for column in text_columns:
    df[column] = df[column].str.strip()

# Standardize text capitalization
for column in text_columns:
    df[column] = df[column].str.title()

# Convert Quantity to numeric
df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

# Convert Sales Amount to numeric
df["Sales_Amount"] = pd.to_numeric(
    df["Sales_Amount"],
    errors="coerce"
)

# Convert Order Date to datetime
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)

# Handle missing text values
df["Customer_Name"] = df["Customer_Name"].fillna("Unknown")
df["City"] = df["City"].fillna("Unknown")
df["Category"] = df["Category"].fillna("Unknown")
df["Product"] = df["Product"].fillna("Unknown")
df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")
df["Order_Status"] = df["Order_Status"].fillna("Unknown")

print("Data cleaning completed!")

display(df.head())
```

```python
# Calculate median Sales Amount
median_sales = df["Sales_Amount"].median()

print("Median Sales Amount:", median_sales)

# Replace missing Sales Amount values
df["Sales_Amount"] = df["Sales_Amount"].fillna(median_sales)

print(
    "\nMissing Sales Amounts after cleaning:",
    df["Sales_Amount"].isnull().sum()
)
```

```python
print("===== FINAL DATASET CHECK =====")

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Cleaned Records:")
display(df.head())

print("\nLast 5 Cleaned Records:")
display(df.tail())
```

```python
cleaned_file = "Friend_Task1_Cleaned_Sales_Dataset.csv"

df.to_csv(cleaned_file, index=False)

print("Cleaned dataset saved successfully!")
print("File name:", cleaned_file)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```
